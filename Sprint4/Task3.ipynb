**Описание задачи:**

Создать базовый класс Subscription в C#, который будет представлять подписки на
различные услуги. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

**Требования к базовому классу Subscription:**

• Атрибуты: ID подписки (SubscriptionId), Название услуги (ServiceName),
Стоимость подписки (Cost).

•Методы:

o CalculateMonthlyCost(): метод для расчета ежемесячной стоимости
подписки.

o ExtendSubscription(): метод для продления подписки на
дополнительный период.

o GetSubscriptionDetails(): метод для получения деталей подписки.

**Требования к производным классам:**
1. ПодпискаНаОнлайнСервис (OnlineServiceSubscription): Должна содержать
дополнительные атрибуты, такие как Количество доступных пользователей
(MaxUsers). Метод CalculateMonthlyCost() должен быть переопределен для
учета количества пользователей при расчете стоимости.

2. ПодпискаНаСтreamинг (StreamingSubscription): Должна содержать
дополнительные атрибуты, такие как Количество одновременных потоков
(MaxStreams). Метод ExtendSubscription() должен быть переопределен для
добавления специальных предложений для продления подписки.

3. ПодпискаНаВидео(VideoSubscription) (если требуется третий класс): Должна
содержать дополнительные атрибуты, такие как Качество видео
(VideoQuality). Метод GetSubscriptionDetails() должен быть переопределен
для отображения качества видео вместе с другими деталями подписки.

```
public class Subscription{ //Базовый класс
    // Основные поля и свойства
    public int SubscriptionId { get; private set; }     // ID подписки
    protected string ServiceName { get; private set; }  // Название сервиса
    private double Cost;                                // Стоимость (инкапсулирована)

    // Свойство с валидацией значения
    public double GetSetCost
    {
        get { return Cost; }
        set
        {
            if (value >= 0)
                Cost = value;
            else
                throw new ArgumentOutOfRangeException("Стоимость не может быть меньше 0!");
        }
    }

    // Дополнительные характеристики
    public int DurationMonths { get; private set; }  // срок подписки (мес)
    public bool IsActive { get; private set; }       // активна ли подписка
    public string UserEmail { get; private set; }    // email пользователя
    public DateTime StartDate { get; private set; }  // дата начала
    public string PaymentMethod { get; private set; } // способ оплаты
    public string Country { get; private set; }       // страна пользователя
    public int RenewalCount { get; private set; }     // количество продлений

    // Конструктор: создаёт новую подписку
    public Subscription(int id, string serviceName, double cost, string userEmail)
    {
        SubscriptionId = id;
        ServiceName = serviceName;
        Cost = cost;
        DurationMonths = 1;
        IsActive = true;
        UserEmail = userEmail;
        StartDate = DateTime.Now;
        PaymentMethod = "Карта";
        Country = "Россия";
        RenewalCount = 0;
    }

    // Отмена подписки
    public void CancelSubscription()
    {
        IsActive = false;
        Console.WriteLine($"Подписка на {ServiceName} отменена.");
    }

    // Изменение способа оплаты
    public void ChangePaymentMethod(string newMethod)
    {
        PaymentMethod = newMethod;
        Console.WriteLine($"Способ оплаты изменён на {PaymentMethod}");
    }

    // Автоматическое продление
    public void RenewAutomatically()
    {
        if (IsActive)
        {
            RenewalCount++;
            Console.WriteLine($"Подписка {ServiceName} автоматически продлена ({RenewalCount}-й раз).");
        }
    }

    // Изменение страны пользователя
    public void UpdateCountry(string newCountry)
    {
        Country = newCountry;
        Console.WriteLine($"Страна пользователя изменена на {Country}");
    }

    // Расчёт месячной стоимости (виртуальный метод — переопределяется)
    public virtual void CalculateMonthlyCost()
    {
        Console.WriteLine($"Месячная стоимость {ServiceName}: {Cost}₽");
    }

    // Продление подписки без скидки
    public virtual void ExtendSubscription(int months)
    {
        DurationMonths += months;
        RenewalCount++;
        Console.WriteLine($"Подписка продлена на {months} мес. Общая длительность: {DurationMonths} мес.");
    }

    // Перегруженный метод продления со скидкой
    public void ExtendSubscription(int months, double discountPercent)
    {
        double finalCost = Cost - Cost * (discountPercent / 100);
        DurationMonths += months;
        RenewalCount++;
        Console.WriteLine($"Продление на {months} мес. со скидкой {discountPercent}% — {finalCost * months}₽");
    }

    // Вывод общей информации о подписке
    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine($"ID: {SubscriptionId}, Сервис: {ServiceName}, Страна: {Country}, Стоимость: {Cost}₽, Активна: {IsActive}");
    }
}

public class OnlineServiceSubscription : Subscription{  //Простое наследование
     private int _maxUsers; // максимальное число пользователей
    public int GetMaxUsers() => _maxUsers; // получить ограничение

    // Дополнительные свойства
    public string AccessLevel { get; private set; }     // уровень доступа (Basic, Pro)
    public DateTime LastLogin { get; private set; }     // дата последнего входа
    public bool HasPrioritySupport { get; private set; } // приоритетная поддержка
    public string[] UserRoles { get; private set; }     // роли пользователей
    public int ActiveSessions { get; private set; }     // активные сессии

    // Проверка корректности количества пользователей
    protected void SetMaxUsers(int value)
    {
        if (value > 0)
            _maxUsers = value;
        else
            throw new ArgumentOutOfRangeException("Количество пользователей не может быть меньше 0!");
    }

    // Конструктор: инициализация онлайн-подписки
    public OnlineServiceSubscription(int id, string name, double cost, string email, int maxUsers)
        : base(id, name, cost, email)
    {
        SetMaxUsers(maxUsers);
        AccessLevel = "Basic";
        LastLogin = DateTime.Now;
        HasPrioritySupport = false;
        UserRoles = new string[] { "Viewer" };
        ActiveSessions = 0;
    }

    // Повышение уровня доступа
    public void UpgradeAccess(string newLevel)
    {
        AccessLevel = newLevel;
        Console.WriteLine($"Уровень доступа изменён: {AccessLevel}");
    }

    // Включение приоритетной поддержки
    public void EnablePrioritySupport()
    {
        HasPrioritySupport = true;
        Console.WriteLine("Приоритетная поддержка активирована.");
    }

    // Добавление новой роли пользователю
    public void AddUserRole(string role)
    {
        var roles = new List<string>(UserRoles);
        roles.Add(role);
        UserRoles = roles.ToArray();
        Console.WriteLine($"Добавлена роль: {role}");
    }

    // Учёт начала или завершения сессии
    public void TrackSession(bool start)
    {
        ActiveSessions += start ? 1 : -1;
        Console.WriteLine($"Активных сессий: {ActiveSessions}");
    }

    // Переопределённый метод расчёта стоимости (полиморфизм)
    public override void CalculateMonthlyCost()
    {
        Console.WriteLine($"Стоимость для {GetMaxUsers()} пользователей: {GetSetCost * GetMaxUsers()}₽");
    }
}

public class StreamingSubscription : Subscription{ //Простое наслдеование от Subscription, но образует сложное наследование, 
//так как от него наследуется PremiumSubscription
    private int _maxStreams; // максимальное число потоков

    // Дополнительные поля
    public bool HasOfflineMode { get; private set; }    // доступен ли оффлайн режим
    public string Region { get; private set; }          // регион пользователя
    public double DiscountPercent { get; private set; } // процент скидки
    public int ContentCount { get; private set; }       // количество фильмов/треков
    public string PreferredQuality { get; private set; } // качество (HD, 4K)

    // Конструктор: создание стриминговой подписки
    public StreamingSubscription(int id, string name, double cost, string email, int maxStreams)
        : base(id, name, cost, email)
    {
        _maxStreams = maxStreams;
        HasOfflineMode = false;
        Region = "Россия";
        DiscountPercent = 0;
        ContentCount = 0;
        PreferredQuality = "HD";
    }

    // Включить оффлайн режим
    public void EnableOfflineMode()
    {
        HasOfflineMode = true;
        Console.WriteLine("Оффлайн-режим активирован.");
    }

    // Изменить качество контента
    public void SetQuality(string quality)
    {
        PreferredQuality = quality;
        Console.WriteLine($"Качество изменено на: {PreferredQuality}");
    }

    // Применить скидку
    public void ApplyDiscount(double percent)
    {
        DiscountPercent = percent;
        Console.WriteLine($"Применена скидка {DiscountPercent}%");
    }

    // Добавить контент (например, новый фильм)
    public void AddContent(int count)
    {
        ContentCount += count;
        Console.WriteLine($"Добавлено {count} единиц контента. Всего: {ContentCount}");
    }

    // Переопределённый расчёт стоимости с учётом скидки
    public override void CalculateMonthlyCost()
    {
        double finalCost = GetSetCost * (1 - DiscountPercent / 100);
        Console.WriteLine($"Итоговая стоимость подписки ({PreferredQuality}): {finalCost}₽");
    }

}

public interface IRewardable
{
    void GiveBonusPoints(int points); // начислить бонусы
}

public interface IDiscountable
{
    void ApplyDiscount(double percent); // применить скидку
}

public class PremiumSubscription : StreamingSubscription, IRewardable, IDiscountable 
//Множественное наследование через интерфейсы, интерфейсы IRewardable и IDiscountable. 
//Класс PremiumSubscription наследует StreamingSubscription - Сложное наследование, так как StreamingSubscription уже является производным от Subscription;
// также PremiumSubscription реализует оба интерфейса IRewardable и IDiscountable.
{
    // Новые атрибуты
    public int BonusPoints { get; private set; }     // бонусные баллы
    public string PremiumLevel { get; private set; } // уровень премиума
    public bool HasGiftIncluded { get; private set; } // наличие подарка
    public string VIPManager { get; private set; }    // персональный менеджер

    // Конструктор
    public PremiumSubscription(int id, string name, double cost, string userEmail, int maxStreams)
        : base(id, name, cost, userEmail, maxStreams)
    {
        BonusPoints = 0;
        PremiumLevel = "Gold";
        HasGiftIncluded = true;
        VIPManager = "Иван Иванов";
    }

    // Начисление бонусов
    public void GiveBonusPoints(int points)
    {
        BonusPoints += points;
        Console.WriteLine($"Начислено {points} бонусов. Всего: {BonusPoints}");
    }

    // Переопределённая скидка (реализация интерфейса)
    public new void ApplyDiscount(double percent)
    {
        GetSetCost -= GetSetCost * (percent / 100);
        Console.WriteLine($"Скидка {percent}% применена. Новая цена: {GetSetCost}₽");
    }

    // Показ преимуществ
    public void ShowPremiumBenefits()
    {
        Console.WriteLine($"Уровень: {PremiumLevel}. Подарок включён: {HasGiftIncluded}. Менеджер: {VIPManager}");
    }

    // Обновление уровня премиума
    public void UpgradeToPlatinum()
    {
        PremiumLevel = "Platinum";
        Console.WriteLine("Премиум-уровень повышен до Platinum!");
    }
}

public class VideoSubscription : Subscription{
    private int _videoQuality;

    // Свойства
    public bool IsHDREnabled { get; private set; } // HDR поддержка
    public string CodecType { get; private set; }   // тип кодека
    public int MaxResolution { get; private set; }  // разрешение
    public double FrameRate { get; private set; }   // частота кадров

    // Конструктор
    public VideoSubscription(int id, string name, double cost, string email, int quality)
        : base(id, name, cost, email)
    {
        _videoQuality = quality;
        IsHDREnabled = false;
        CodecType = "H.264";
        MaxResolution = quality;
        FrameRate = 30;
    }

    // Включение HDR
    public void EnableHDR()
    {
        IsHDREnabled = true;
        Console.WriteLine("HDR режим включён.");
    }

    // Изменение кодека
    public void ChangeCodec(string codec)
    {
        CodecType = codec;
        Console.WriteLine($"Кодек изменён на {CodecType}");
    }

    // Обновление разрешения
    public void UpgradeResolution(int newRes)
    {
        if (newRes > MaxResolution)
        {
            MaxResolution = newRes;
            Console.WriteLine($"Разрешение увеличено до {MaxResolution}p");
        }
        else
        {
            Console.WriteLine("Новое разрешение не выше текущего.");
        }
    }

    // Изменение частоты кадров
    public void SetFrameRate(double fps)
    {
        FrameRate = fps;
        Console.WriteLine($"Частота кадров установлена: {FrameRate} FPS");
    }

    // Переопределение метода вывода информации
    public override void GetSubscriptionDetails()
    {
        Console.WriteLine($"ID: {SubscriptionId}, Сервис: {ServiceName}, Качество: {MaxResolution}p, Кодек: {CodecType}, FPS: {FrameRate}");
    }
}

// Универсальный менеджер подписок (работает с любым типом наследником Subscription)
public class SubscriptionManager<T> where T : Subscription
{
    private List<T> subscriptions = new List<T>();

    // Добавление подписки
    public void AddSubscription(T sub)
    {
        subscriptions.Add(sub);
        Console.WriteLine($"Добавлена подписка: {sub.GetType().Name}");
    }

    // Удаление по ID
    public void RemoveById(int id)
    {
        var sub = subscriptions.Find(s => s.SubscriptionId == id);
        if (sub != null)
        {
            subscriptions.Remove(sub);
            Console.WriteLine($"Подписка ID {id} удалена.");
        }
        else
        {
            Console.WriteLine($"Подписка ID {id} не найдена.");
        }
    }

    // Показ всех подписок
    public void ShowAll()
    {
        Console.WriteLine("Текущие подписки:");
        foreach (var s in subscriptions)
        {
            s.GetSubscriptionDetails();
        }
    }
}

// НОВЫЙ ИНТЕРФЕЙС
public interface ISubscriptionService
{
    void Activate();
    void Suspend();
    void DisplayStatus();
}

// КЛАСС С ЯВНОЙ РЕАЛИЗАЦИЕЙ ИНТЕРФЕЙСА 
public class PremiumControlService : ISubscriptionService
{
    private readonly PremiumSubscription _premium;

    public PremiumControlService(PremiumSubscription premium)
    {
        _premium = premium;
    }

    // Явная реализация интерфейса
    void ISubscriptionService.Activate()
    {
        Console.WriteLine($"Активация премиум-подписки {_premium.GetType().Name}");
        _premium.GiveBonusPoints(50); // бонус за активацию
    }

    void ISubscriptionService.Suspend()
    {
        _premium.CancelSubscription();
        Console.WriteLine("Премиум-подписка временно приостановлена.");
    }

    void ISubscriptionService.DisplayStatus()
    {
        _premium.GetSubscriptionDetails();
        _premium.ShowPremiumBenefits();
    }
}

// КЛАСС, УПРАВЛЯЮЩИЙ ЗАВИСИМОСТЬЮ 
public class SubscriptionController
{
    private readonly ISubscriptionService _service;

    // Внедрение зависимости через конструктор
    public SubscriptionController(ISubscriptionService service)
    {
        _service = service;
    }

    public void RunSubscriptionFlow()
    {
        _service.Activate();
        _service.DisplayStatus();
        _service.Suspend();
    }
}

Console.WriteLine("Демонстрация явной реализации интерфейса и внедрения зависимости:");
Console.WriteLine();

PremiumSubscription premiumSub = new PremiumSubscription(10, "UltraVideo", 3500, "vip@mail.com", 3);

// Явная реализация интерфейса
ISubscriptionService service = new PremiumControlService(premiumSub);

// Внедрение зависимости
SubscriptionController controller = new SubscriptionController(service);


// Запуск сценария работы
controller.RunSubscriptionFlow();
Console.WriteLine();
```

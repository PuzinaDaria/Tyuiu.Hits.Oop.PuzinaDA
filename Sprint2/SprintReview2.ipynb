```
public class Employee //Базовый класс Employee (Сотрудник) -  основной класс для всех типов сотрудников

{
    // Свойства сотрудника
    public string Name { get; set; }           // Имя сотрудника
    public string Position { get; set; }       // Должность сотрудника
    public double Salary { get; set; }         // Зарплата сотрудника
   
    public Employee(string name, string position, double salary)  // Создание нового сотрудника
    {
        Name = name;          // Устанавливаем имя
        Position = position;  // Устанавливаем должность
        Salary = salary;      // Устанавливаем зарплату
    }

    //Вывод информации о сотруднике в консоль.
    public virtual void PrintInfo() // Виртуальный метод для того, чтобы можно было переопределить в производных классах
    {
        Console.WriteLine($"Сотрудник: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Зарплата: {Salary} руб.");
    }
}
 
public class Manager : Employee //Простое наследование - Класс Manager, наследуется от Employee и добавляет функциональность менеджера
{
    //Отдел управления
    public string Department { get; set; }

    // Конструктор
    public Manager(string name, string position, double salary, string department) 
        : base(name, position, salary)  // Вызывает конструктор базового класса, передает параметры в базовый класс
    {
        Department = department;  // Устанавливаем отдел
    }

    
    public override void PrintInfo() // Переопределяем метод - полиморфизм
    {
        // Выводим базовую информацию
        base.PrintInfo();
        // Добавляем информацию об отделе
        Console.WriteLine($"Отдел: {Department}");
    }
}

public class ProjectManager : Manager //Наследуется от Manager (Сложное наследование) и добавляет функциональность проектного менеджера
{
    
    public string ProjectName { get; set; } // Название проекта

   
    public ProjectManager(string name, string position, double salary, string department, string projectName) // Название проекта - дополнительное свойство
        : base(name, position, salary, department)  // Передаем параметры в класс Manager
    {
        ProjectName = projectName;  // Устанавливаем название проекта
    }

    
    public override void PrintInfo() //Переопредление метода - добавляем информацию о проекте
    {
        // Вызываем метод из класса Manager (Manager вызывает метод из Employee)
        base.PrintInfo();
        // Добавляем информацию о проекте
        Console.WriteLine($"Проект: {ProjectName}");
    }
}

public interface IReportable //Интерфейс IReportable  - для генерации отчетов. Интерфейс определяет контракт для классов, которые могут генерировать отчеты
{
    string GenerateReport();  // Метод должен возвращать строку с отчетом
}

public interface ITeamLeader // Интерфейс ITeamLeader - для управления командой. Интерфейс определяет контракт для классов, которые управляют командой
{
    string[] GetTeamMembers();  // Метод должен возвращать массив членов команды
}


public class AdvancedProjectManager : ProjectManager, IReportable, ITeamLeader // Множественное наследование, наследует ProjectManager и реализует два интерфейса
{
    // Члены команды
    private string[] teamMembers;

    // Конструктор
    public AdvancedProjectManager(string name, string position, double salary, string department, string projectName, string[] team) 
        : base(name, position, salary, department, projectName)
    {
        teamMembers = team;  // Сохраняем массив членов команды
    }

    
    public string GenerateReport() //реализация метода из интерфейса IReportable 
    {
        // Генерируем отчет с информацией о проекте и команде
        return $"Отчет по проекту '{ProjectName}'. Команда: {teamMembers.Length} человек. Статус: В работе";
    }

    
    public string[] GetTeamMembers() //реализация метода из интерфейса ITeamLeader
    {
        // Возвращаем массив членов команды
        return teamMembers;
    }
    public override void PrintInfo() //Переопределение метода - добавляем информацию о команде
    {
        // Вызываем метод из ProjectManager (Который вызывает из Manager, а Manager вызывает из Employee)
        base.PrintInfo();
        // Добавляем информацию о команде
        Console.WriteLine($"Размер команды: {teamMembers.Length} человек");
    }
}


Console.WriteLine("Система управления сотрудниками");
Console.WriteLine();

// Создаем массив членов команды для проектного менеджера
string[] teamMembers = { "Иван Иванов", "Петр Петрович", "Ирина Васильева", "Сергей Сергеев" };

// Создаем различных сотрудников
Employee employee = new Employee("Иван Кирилов", "Разработчик", 48000);
Manager manager = new Manager("Катерина Смирнова", "Менеджер отдела", 75000, "IT отдел");
ProjectManager projectManager = new ProjectManager("Тимофей Сидоров", "Проектный менеджер", 110000, "Разработка", "Новое приложение");
AdvancedProjectManager advancedPM = new AdvancedProjectManager("София Зайцева", "Старший проектный менеджер", 115000, "Разработка", "Мобильное приложение", teamMembers);

// Тестируем базового сотрудника
Console.WriteLine(" Базовый сотрудник: ");
employee.PrintInfo();
Console.WriteLine();

// Тестируем менеджера
Console.WriteLine(" Менеджер: ");
manager.PrintInfo();
Console.WriteLine();

// Тестируем проектного менеджера
Console.WriteLine(" Проектный менеджер: ");
projectManager.PrintInfo();
Console.WriteLine();

// Тестируем расширенного проектного менеджера (множественное наследование)
Console.WriteLine(" Продвинутый проектный менеджер: ");
advancedPM.PrintInfo();

// Тестируем методы интерфейсов
Console.WriteLine();
Console.WriteLine("Тестирование интерфейсов");

// Тестируем генерацию отчета (интерфейс IReportable)
Console.WriteLine($"Отчет: {advancedPM.GenerateReport()}");

// Тестируем получение членов команды (интерфейс ITeamLeader)
string[] team = advancedPM.GetTeamMembers();
Console.WriteLine("Члены команды:");
foreach (string member in team) //временная переменная для каждого члена команды и перебираем в массиве team
{
    Console.WriteLine(member); // Выводим всех членов команды
}
```

**Описание задачи:**

Создать базовый класс Subscription в C#, который будет представлять подписки на
различные услуги. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

**Требования к базовому классу Subscription:**

• Атрибуты: ID подписки (SubscriptionId), Название услуги (ServiceName),
Стоимость подписки (Cost).

•Методы:

o CalculateMonthlyCost(): метод для расчета ежемесячной стоимости
подписки.

o ExtendSubscription(): метод для продления подписки на
дополнительный период.

o GetSubscriptionDetails(): метод для получения деталей подписки.

**Требования к производным классам:**
1. ПодпискаНаОнлайнСервис (OnlineServiceSubscription): Должна содержать
дополнительные атрибуты, такие как Количество доступных пользователей
(MaxUsers). Метод CalculateMonthlyCost() должен быть переопределен для
учета количества пользователей при расчете стоимости.

2. ПодпискаНаСтreamинг (StreamingSubscription): Должна содержать
дополнительные атрибуты, такие как Количество одновременных потоков
(MaxStreams). Метод ExtendSubscription() должен быть переопределен для
добавления специальных предложений для продления подписки.

3. ПодпискаНаВидео(VideoSubscription) (если требуется третий класс): Должна
содержать дополнительные атрибуты, такие как Качество видео
(VideoQuality). Метод GetSubscriptionDetails() должен быть переопределен
для отображения качества видео вместе с другими деталями подписки.

```
using System;
using System.Collections.Generic;

/// Делегат, добавленный для реализации системы событий.
/// Передаёт объект подписки и текст сообщения.
public delegate void SubscriptionEventHandler(Subscription sub, string message);

public class Subscription
{
    public int SubscriptionId { get; private set; }
    public string ServiceName { get; set; }
    public double GetSetCost { get; set; }
    public bool IsActive { get; private set; } = true;
    public string UserEmail { get; set; }
    public string PaymentMethod { get; private set; }
    public int DurationMonths { get; protected set; }
    public int RenewalCount { get; private set; } = 0;
    public string Country { get; private set; }

    /// Событие, добавленное для отслеживания изменений подписки:
    /// продления, отмены, смены оплаты, смены страны и т.п.
    public event SubscriptionEventHandler OnSubscriptionChanged;

    public Subscription(int id, string name, double cost, string userEmail)
    {
        SubscriptionId = id;
        ServiceName = name;
        GetSetCost = cost;
        UserEmail = userEmail;
        PaymentMethod = "Not Specified";
        DurationMonths = 1;
        Country = "Unknown";
    }

    public double CalculateTotalCost(int months)
    {
        return GetSetCost * months;
    }

    public void UpdatePaymentMethod(string newMethod)
    {
        PaymentMethod = newMethod;
        Console.WriteLine($"Способ оплаты обновлён: {newMethod}");

        // вызывается событие изменения подписки
        OnSubscriptionChanged?.Invoke(this, "Изменён способ оплаты");
    }

    public void CancelSubscription()
    {
        IsActive = false;
        Console.WriteLine($"Подписка на {ServiceName} отменена.");

        // уведомление о событии отмены
        OnSubscriptionChanged?.Invoke(this, "Подписка отменена");
    }

    public virtual void ExtendSubscription(int months)
    {
        DurationMonths += months;
        RenewalCount++;
        Console.WriteLine($"Подписка продлена на {months} мес.");

        // уведомление о продлении
        OnSubscriptionChanged?.Invoke(this, $"Продлена на {months} мес.");
    }

    public void ExtendSubscription(int months, double discountPercent)
    {
        double finalCost = GetSetCost - GetSetCost * (discountPercent / 100);
        DurationMonths += months;
        RenewalCount++;

        Console.WriteLine($"Продление: {months} мес. скидка {discountPercent}%, сумма {finalCost * months}₽");

        // уведомление о продлении со скидкой
        OnSubscriptionChanged?.Invoke(this, $"Продлена со скидкой {discountPercent}%");
    }

    public void UpdateCountry(string newCountry)
    {
        Country = newCountry;
        Console.WriteLine($"Страна обновлена: {Country}");

        // уведомление о смене страны
        OnSubscriptionChanged?.Invoke(this, $"Страна пользователя изменена на {Country}");
    }

    public void GetInfo()
    {
        Console.WriteLine($"ID: {SubscriptionId}, Сервис: {ServiceName}, Цена: {GetSetCost}₽");
    }
}

public class StreamingSubscription : Subscription
{
    public int MaxDevices { get; private set; }
    public int ContentCount { get; private set; }

    /// Событие добавлено для уведомлений о появлении нового контента.
    public event SubscriptionEventHandler OnNewContentAdded;

    public StreamingSubscription(int id, string name, double cost, string userEmail, int maxDevices)
        : base(id, name, cost, userEmail)
    {
        MaxDevices = maxDevices;
        ContentCount = 0;
    }

    public void AddContent(int count)
    {
        ContentCount += count;
        Console.WriteLine($"Добавлено {count} элементов контента.");

        // вызывается событие появления нового контента
        OnNewContentAdded?.Invoke(this, $"{count} новых элементов");
    }
}

public class PremiumSubscription : StreamingSubscription
{
    public int BonusPoints { get; private set; }
    public string PremiumLevel { get; private set; }
    public bool HasGiftIncluded { get; private set; }
    public string VIPManager { get; private set; }
    /// Коллекция добавлена для хранения списка подарков премиум-подписки.
    public List<string> Gifts { get; private set; }

    /// Событие добавлено для уведомлений о начислении бонусных баллов.
    public event SubscriptionEventHandler OnBonusAdded;

    public PremiumSubscription(int id, string name, double cost, string userEmail, int maxStreams)
        : base(id, name, cost, userEmail, maxStreams)
    {
        BonusPoints = 0;
        PremiumLevel = "Gold";
        HasGiftIncluded = true;
        VIPManager = "Иван Иванов";

        // список подарков премиум-подписки
        Gifts = new List<string> { "Стикер", "Купон на месяц" };
    }

    public void GiveBonusPoints(int points)
    {
        BonusPoints += points;
        Console.WriteLine($"Добавлено {points} бонусов.");

        // событие начисления бонусов
        OnBonusAdded?.Invoke(this, $"Начислено {points} бонусных баллов");
    }
}

public class SubscriptionManager<T> where T : Subscription
{
    private List<T> subscriptions = new List<T>();

    /// Добавлен журнал событий, куда записываются все уведомления.
    private List<string> HistoryLog = new List<string>();

    public void AddSubscription(T sub)
    {
        subscriptions.Add(sub);

        // подключение подписки к событиям всех типов
        SubscribeToEvents(sub);

        Console.WriteLine($"Добавлена подписка: {sub.ServiceName}");
    }

    /// Подключение обработчиков событий — метод добавлен.
    /// Реагирует на события обычных, стриминговых и премиальных подписок.
    private void SubscribeToEvents(T sub)
    {
        sub.OnSubscriptionChanged += LogEvent;

        if (sub is StreamingSubscription stream)
            stream.OnNewContentAdded += LogEvent;

        if (sub is PremiumSubscription premium)
            premium.OnBonusAdded += LogEvent;
    }

    /// Метод добавлен для сохранения событий в журнал.
    private void LogEvent(Subscription sub, string message)
    {
        string log = $"[{DateTime.Now}] {sub.ServiceName}: {message}";
        HistoryLog.Add(log);
        Console.WriteLine("LOG: " + log);
    }

    public void ShowHistory()
    {
        Console.WriteLine("\nЖурнал событий:");
        foreach (var entry in HistoryLog)
            Console.WriteLine(entry);
    }
}


// пример использования, который создаёт такой вывод:
var mgr = new SubscriptionManager<Subscription>();

var premium = new PremiumSubscription(1, "VideoMax", 2500, "vip@mail.com", 3);
var stream = new StreamingSubscription(2, "CineStream", 900, "cine@mail.com", 2);

mgr.AddSubscription(premium);
mgr.AddSubscription(stream);

Console.WriteLine();

premium.ExtendSubscription(3);
premium.GiveBonusPoints(200);
Console.WriteLine();

stream.AddContent(5);
stream.UpdatePaymentMethod("PayPal");
Console.WriteLine();

premium.CancelSubscription();

mgr.ShowHistory();
```
